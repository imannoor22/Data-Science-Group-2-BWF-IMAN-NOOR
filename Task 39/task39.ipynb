### What is RNN?
Recurrent Neural Networks (RNNs) are a class of neural networks designed for processing sequential data. They maintain a hidden state that captures information about previous inputs, making them suitable for tasks such as language modeling and time series prediction.

### Variants of RNN

#### Long Short-Term Memory (LSTM)
LSTMs are a type of RNN that are capable of learning long-term dependencies. They address the vanishing gradient problem that standard RNNs face by using a special gating mechanism to control the flow of information. This allows LSTMs to remember information for longer periods, making them effective for tasks like language translation and speech recognition.

#### Gated Recurrent Unit (GRU)
GRUs are another variant of RNNs that simplify the LSTM architecture by combining the forget and input gates into a single update gate. This makes GRUs faster to train and often just as effective as LSTMs for many tasks. They are also used in various applications involving sequential data.

### Steps:

### 1. Collect Data

- **Data Collection:** Use a small set of text, such as sentences from a book or articles. For simplicity, you can start with a dataset of short sentences or paragraphs.

### 2. Prepare Data

- **Data Preparation:** Convert sentences into sequences of words or characters. Tokenize the text and create sequences of a fixed length. The target for each sequence will be the next word following the sequence.

### 3. Build Model

- **Model Architecture:** Create an RNN-based model that learns to predict the next word in a sequence based on the previous words. The basic RNN can be enhanced with Long Short-Term Memory (LSTM) or Gated Recurrent Unit (GRU) cells for better performance.

### 4. Generate Text

- **Text Generation:** Once the model is trained, input a few words to seed the text generation process. The model will predict the next word, and you can iteratively generate a sequence by feeding the predicted word back into the model.

### Purpose:
This basic task helps understand how RNNs, LSTMs, and GRUs handle sequential data and make predictions based on learned patterns.


```python
import numpy as np
import regex as re
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:#ff7f50; font-weight:bold;">
    DATA COLLECTION
</h1>

```python
def file_to_sentence_list(file_path): 
    with open(file_path, 'r') as file: 
        text = file.read() 
  
    # Splitting the text into sentences using 
    # delimiters like '.', '?', and '!' 
    sentences = [sentence.strip() for sentence in re.split( 
        r'(?<=[.!?])\s+', text) if sentence.strip()] 
  
    return sentences
```

```python
file_path = '/kaggle/input/wiki-data/wiki-text.txt'
text_data = file_to_sentence_list(file_path) 
```

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:#ff7f50; font-weight:bold;">
    DATA PREPARATION
</h1>

```python
# Tokenize the text data 
tokenizer = Tokenizer() 
tokenizer.fit_on_texts(text_data) 
total_words = len(tokenizer.word_index) + 1
  
# Create input sequences 
input_sequences = [] 
for line in text_data: 
    token_list = tokenizer.texts_to_sequences([line])[0] 
    for i in range(1, len(token_list)): 
        n_gram_sequence = token_list[:i+1] 
        input_sequences.append(n_gram_sequence) 
        
# Pad sequences and split into predictors and label 
max_sequence_len = max([len(seq) for seq in input_sequences]) 
input_sequences = np.array(pad_sequences( 
    input_sequences, maxlen=max_sequence_len, padding='pre')) 
X, y = input_sequences[:, :-1], input_sequences[:, -1] 
  
# Convert target data to one-hot encoding 
y = tf.keras.utils.to_categorical(y, num_classes=total_words) 
```

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:#ff7f50; font-weight:bold;">
    BUILDING MODEL
</h1>

```python
model = Sequential()
model.add(Embedding(total_words, 10, input_length=max_sequence_len-1))
model.add(LSTM(128))
model.add(Dense(total_words, activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

## **Training the model**

```python
model.fit(X, y, epochs=100, verbose=1)
```

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:#ff7f50; font-weight:bold;">
    GENERATING TEXT
</h1>

## **Predicting the next word**

```python
# Generate next word predictions 
seed_text = "York mostly occupied "
next_words = 10
  
for _ in range(next_words): 
    token_list = tokenizer.texts_to_sequences([seed_text])[0] 
    token_list = pad_sequences( 
        [token_list], maxlen=max_sequence_len-1, padding='pre') 
    predicted_probs = model.predict(token_list) 
    predicted_word = tokenizer.index_word[np.argmax(predicted_probs)] 
    seed_text += " " + predicted_word 
  
print("Next predicted words:", seed_text)
```
