    NLP PREPROCESSING
</h1>

Natural Language Processing (NLP) preprocessing is a crucial step in preparing textual data for analysis and model building. It involves a series of techniques to clean and transform raw text into a format that can be effectively processed by machine learning algorithms. The primary goal of NLP preprocessing is to reduce noise, standardize text, and extract meaningful features that can improve the performance of NLP models.

Key steps in NLP preprocessing include:

1. **Tokenization**: Splitting text into individual units, such as words or sentences, which serve as the building blocks for further analysis.

2. **Lowercasing**: Converting all text to lowercase to ensure uniformity and reduce the complexity of the vocabulary.

3. **Removing Punctuation**: Eliminating punctuation marks to focus on the words themselves and avoid unnecessary symbols in the analysis.

4. **Stopword Removal**: Removing common words (e.g., "and", "the", "is") that do not carry significant meaning and may clutter the data.

5. **Stemming and Lemmatization**: Reducing words to their base or root form. Stemming chops off prefixes and suffixes, while lemmatization considers the word's context and reduces it to its dictionary form.

6. **Removing Special Characters**: Eliminating non-alphanumeric characters, such as hashtags, mentions, or emojis, to clean the text further.

7. **Handling Numerical Data**: Converting or removing numbers in the text depending on their relevance to the task.

8. **Text Normalization**: Standardizing text by expanding contractions (e.g., "can't" to "cannot"), correcting misspellings, and normalizing word forms.

9. **Removing Rare Words or Characters**: Filtering out words or characters that appear infrequently, which may not contribute significantly to the analysis.

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:navy; text-align:center;">
    TOKENIZATION
</h1>


Tokenization is a fundamental step in the preprocessing of text data in Natural Language Processing (NLP). It involves breaking down a stream of text into smaller, manageable units called tokens. Tokens can be words, sentences, or even subwords, depending on the level of tokenization required for the task.

#### Types of Tokenization:

1. **Word Tokenization**: 
   - The text is split into individual words. This is the most common type of tokenization and is used in tasks where each word is treated as a separate unit.
   - Example: The sentence "Natural Language Processing is fascinating!" would be tokenized into ["Natural", "Language", "Processing", "is", "fascinating", "!"].

2. **Sentence Tokenization**:
   - The text is split into sentences. This type of tokenization is useful in tasks where sentence structure or the relationship between sentences is important.
   - Example: The paragraph "I love NLP. It's a very interesting field." would be tokenized into ["I love NLP.", "It's a very interesting field."].

3. **Subword Tokenization**:
   - The text is split into subword units, which can be useful for handling out-of-vocabulary words or languages with complex morphology.
   - Example: The word "unhappiness" might be tokenized into ["un", "happiness"].

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:navy; text-align:center;">
    LOWERCASING
</h1>

Lowercasing is a simple yet essential preprocessing step in Natural Language Processing (NLP) that involves converting all the characters in a text to lowercase. This process helps standardize the text by reducing the variability caused by different cases (uppercase vs. lowercase) and ensures that words are treated uniformly, regardless of how they are capitalized in the original text.

#### Importance of Lowercasing:

1. **Uniformity in Text**:
   - Lowercasing helps create a consistent text format by ensuring that words like "Natural", "NATURAL", and "natural" are all treated as the same word. This uniformity is crucial for reducing the dimensionality of the text data.

2. **Improved Model Performance**:
   - By reducing the number of distinct words (i.e., different capitalizations of the same word), lowercasing can help improve the efficiency and performance of NLP models. It allows models to learn more effectively by focusing on the semantic meaning of words rather than their case variations.

3. **Simplicity in Matching**:
   - Lowercasing simplifies tasks like matching, searching, and comparing words. For example, in text search applications, converting all text to lowercase ensures that searches are case-insensitive, providing more accurate results.

4. **Handling Ambiguity**:
   - In some languages and contexts, the case of a word can change its meaning. However, in many NLP tasks, such as sentiment analysis or topic modeling, lowercasing is sufficient and can help avoid confusion caused by case variations.

#### When to Avoid Lowercasing:

- **Proper Nouns and Acronyms**: In tasks where proper nouns, acronyms, or case-sensitive information (e.g., "USA" vs. "usa") are important, lowercasing may not be appropriate, as it could lead to the loss of critical information.
- **Context-Sensitive Tasks**: For tasks that rely on case distinctions, such as Named Entity Recognition (NER) or certain forms of syntactic parsing, lowercasing might need to be applied selectively or not at all.


```python
def lowercase_text(text):
    return text.lower()
```

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:navy; text-align:center;">
    STEMMING
</h1>

It involves reducing words to their root or base form by removing suffixes, prefixes, or other inflected forms. The primary goal of stemming is to treat words with the same root as equivalent, which helps in simplifying the text and reducing the dimensionality of the data.

#### How Stemming Works:

- Stemming algorithms, such as the **Porter Stemmer** or **Snowball Stemmer**, apply heuristic rules to strip suffixes and prefixes from words, often producing a root form that may not always be a valid word in the language.
- For example, the words "running", "runner", and "ran" might all be reduced to the stem "run". Similarly, "happiness" and "happy" might be reduced to "happi".

#### Considerations:

- **Crudeness of Stemming**:
   - Stemming is often considered a crude method because it relies on simple rules that can sometimes produce stems that are not actual words (e.g., "studies" might be stemmed to "studi"). This can lead to the loss of linguistic nuances.
  
- **Overstemming and Understemming**:
   - **Overstemming** occurs when a stemming algorithm reduces words to the same stem incorrectly (e.g., "universe" and "university" might be reduced to the same stem).
   - **Understemming** happens when words that should be reduced to the same stem are not (e.g., "connect" and "connected" might not be reduced to the same root).

- **When to Use Stemming**:
   - Stemming is most useful in scenarios where exact word forms are not critical, and the focus is on capturing the broader meaning of the text. It is commonly used in applications like search engines, text classification, and topic modeling.

```python
from nltk.stem import PorterStemmer
```

<h1 style="font-size:25px; font-family:Verdana, sans-serif; color:navy; text-align:center;">
    LEMMATIZATION
</h1>

Lemmatization is a text preprocessing technique in Natural Language Processing (NLP) that reduces words to their base or root form, known as the "lemma." Unlike stemming, which simply chops off prefixes or suffixes, lemmatization considers the context and morphological structure of the word, resulting in a valid word that belongs to the language.

#### How Lemmatization Works:

- Lemmatization uses linguistic knowledge, such as part-of-speech tags and the word's meaning, to determine the correct base form of a word. For example, the words "running", "ran", and "runs" are all reduced to the lemma "run".
- Lemmatization requires a detailed dictionary or a database that maps various word forms to their corresponding lemmas.

#### Examples:

- "Studies", "studying", and "studied" are all lemmatized to "study".
- "Children" is lemmatized to "child".
- "Was" and "were" are lemmatized to "be".

#### Considerations:

- **Complexity and Resource Intensity**:
   - Lemmatization is more computationally intensive than stemming because it requires a comprehensive understanding of the language's morphology and context. This complexity can lead to longer processing times, especially for large datasets.

- **Dependency on Accurate Part-of-Speech Tagging**:
   - Lemmatization often relies on accurate part-of-speech tagging to determine the correct lemma. For example, "running" could be a verb ("run") or a noun ("running"), and the correct lemma depends on its usage in the sentence.

- **Applications**:
   - Lemmatization is particularly useful in applications where the meaning and grammatical integrity of words are important, such as machine translation, information retrieval, and advanced text analytics.

```python
# nltk.download('averaged_perceptron_tagger')
# nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    words = word_tokenize(text)
    return [lemmatizer.lemmatize(word) for word in words]

formal_df['lemmatized'] = formal_df['Content'].apply(lambda x: ' '.join(lemmatize_text(x)))
```
